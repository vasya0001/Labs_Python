```python
import pandas as pd
import numpy as np
import seaborn as sn
import matplotlib.pyplot as plt
df = pd.DataFrame()
df['a'] = np.random.rand(100)
df['b'] = np.random.rand(100)
df['c'] = np.random.randint(0,5,100)
df.head()
plt.figure(figsize=(8, 4))
sn.countplot(data=df, x='c', palette='viridis')
plt.title('  Countplot столбца c')
plt.xlabel('Значение c')
plt.ylabel('количество')
plt.show()
sn.displot(data=df,x ='a',color='blue', kind='hist', kde=True) #гистограмма для а,синяя, с прямой
plt.title('Displot для столбца a')
plt.show()
sn.displot(data=df, x = 'b', kind='hist', color='red',kde=True ) #гистограмма для b, с прямрй, красная
plt.title('Displot для столбца b')
plt.show()
sn.displot(data=df,x = 'c', color='green', kde=True, kind = 'hist')#гистограмма для с,с прямой,зеленая
plt.title("Displot для столбца c")
plt.show()
```

```python
import matplotlib.pyplot as plt
labels = ['группа 1', 'группа 2', 'группа 3', 'группа 4', 'группа 5']
men_means = [20, 34, 30, 35, 27]
women_means = [25, 32, 34, 20, 25]
x = range(len(labels))
width = 0.3
plt.bar([i - width/2 for i in x], men_means, width, label='мужчины', color='blue')
plt.bar([i + width/2 for i in x], women_means, width, label='женщины', color='pink')
plt.xlabel('Группы')
plt.ylabel('значения')
plt.title('сравнение показателей мужчин и женщин')
plt.xticks(x, labels)
plt.legend()
plt.show()
```

Задание 8

```python
import matplotlib.pyplot as plt
from sklearn import datasets
data = datasets.load_iris(return_X_y=False)
X = data.data
y = data.target
names = data.target_names
feature_names = data.feature_names
colors = ['red', 'green', 'blue']
fig, axes = plt.subplots(2, 2, figsize=(14, 10)) #создание поля и массива с графиками
axes = axes.flatten() #переаод массива в обычный список
for i in range(4):
    for class_id, color in enumerate(colors):
        mask = (y == class_id) #фильтр нужныъ видов
        axes[i].scatter(
            range(len(X[mask])), X[mask, i],c=color,label=names[class_id])
    axes[i].set_title(feature_names[i], fontsize=14)
    axes[i].set_xlabel('Номер образца', fontsize=10)
    axes[i].set_ylabel('значение', fontsize=10)
    axes[i].grid(True, linestyle='--', alpha=0.5)
    axes[i].legend()
plt.suptitle('Признаки цветов ириса', fontsize=16, y=1.02)
plt.tight_layout() #подгон отступов
plt.show()
```

Задание 9


```python
import matplotlib.pyplot as plt
from sklearn import datasets
data = datasets.load_iris(return_X_y=False)
X = data.data
y = data.target
names = data.target_names
feature_names = data.feature_names
first_feature = [row[0] for row in X]#список из 150 значений
plt.figure(figsize=(10, 6))
colors = ['red', 'green', 'blue']
for class_id in range(3):#строим гистограмму для каждого класса
    class_data = []
    for j in range(len(y)):
        if y[j] == class_id:
            class_data.append(first_feature[j])
    plt.hist(class_data,bins='auto',  alpha=0.5,  color=colors[class_id], label=names[class_id],linewidth=0.5)
plt.title(f'распределение признака: {feature_names[0]}', fontsize=14, fontweight='bold')
plt.xlabel(feature_names[0], fontsize=12)
plt.ylabel('Частота цветков', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--')
plt.tight_layout()
plt.show()


```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn import datasets

data = datasets.load_iris(return_X_y=False)
X = data.data
y = data.target
names = data.target_names
feature_names = data.feature_names
df = pd.DataFrame(X, columns=feature_names)
correlation_matrix = df.corr() #матрица корреляции
#heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm',center=0,vmin=-1, vmax=1,square=True,linewidths=0.5,
cbar_kws={'shrink': 0.8, 'label': 'Коэффициент корреляции'}  # настройки цветовой шкалы
)
plt.title('матрица корреляции признаков цветов ириса', fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
print("матрица корреляции:")
print(correlation_matrix)
```
